The F1-score is defined by the formula
$$
F_1=2\frac{\text{Precision}\times \text{Recall}}{\text{Precision} + \text{Recall}}.
$$

It has essentially weighs both precision and recall equally. However, for this project, it might make more sense to weight recall more heavily.
